```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('../data/breast-cancer.csv')
```

```python
# Muestra todas las columnas sin importar cuántas sean
pd.set_option('display.max_columns', None)
```

```python
df.head()
```

```python
# primera accion binarizar la columna diagnosis y cambiar su nombre a target

df['target']=(df['diagnosis'] == "M").astype(int)
```

```python
df
```

```python
df.describe()
```

```python
#borrar la columna diagnosis ya que ahora vamos a usar 'target'
df.drop('diagnosis',axis=1,inplace=True)
```

```python
df
```

```python
# borrar columnas que no seran utiliez
df
```

```python
# crear la matriz y el vector
X= df.drop('target',axis=1)
y= df['target']
```

```python
# datos de entrenamiento, prubeba, barajear
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.33, random_state=42)
```

```python
#una vez con los datos divididos podemos transformarlos
from sklearn.preprocessing import StandardScaler
scaler= StandardScaler()
X_train= scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# finalmente, luego de normalizarlos a entrenar
from sklearn.svm import SVC
#iterar el modelo
svc_model=SVC()
```

```python
svc_model.fit(X_train,y_train)
```

```python
#evaluar el modelo
predictions = svc_model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(confusion_matrix(y_test,predictions))
```

```python
print(classification_report(y_test,predictions))
```
